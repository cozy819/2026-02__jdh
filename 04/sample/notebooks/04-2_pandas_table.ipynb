# 4-2 pandas와 DataFrame 확인하기
dict가 표가 되고, 같은 열 이름이 그래프로 이어지는 순서를 확인합니다.

```python
data = {
    "day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "visitors": [18, 25, 21, 32, 29],
}
print(data)
```

```python
import pandas as pd

df = pd.DataFrame(data)
display(df)
print("열 이름:", list(df.columns))
print("최댓값:", df["visitors"].max())
```

```python
from pathlib import Path
import matplotlib.pyplot as plt

ax = df.plot(x="day", y="visitors", kind="bar", legend=False, color="#158790", figsize=(8, 4.5))
ax.set_ylim(0, 36)
ax.set_title("Visitors by Day")
ax.set_xlabel("Day")
ax.set_ylabel("Visitors (people)")
ax.bar_label(ax.containers[0])
plt.tight_layout()
output = Path("../output/04-2_dataframe_bar.png")
output.parent.mkdir(exist_ok=True)
plt.savefig(output, dpi=160)
plt.show()
```

## 확인 순서
행 수와 열 이름 → 같은 행의 값 → 최댓값 32 → 사용한 열 → 막대 수와 높이 → 제목·축·단위를 차례로 대조합니다.
