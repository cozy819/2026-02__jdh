4-1에서 Excel로 본 자료와 **내용이 같은 CSV 파일**을 사용합니다. 셀을 위에서 아래로 하나씩 실행합니다.

## 1. 같은 자료, 다른 파일 형식

파일: `../data/04-library-data.csv`

첫 줄은 `date,visitors,borrowed_books`입니다. CSV는 한 줄이 데이터 한 행이고, 값 사이는 쉼표로 나뉩니다.

## 2. 표를 다루는 도구를 가져옵니다

수업 앞부분에서 터미널에 `uv add pandas`를 실행하고 `pyproject.toml`의 변화를 확인했습니다. 아래 셀은 준비한 pandas를 코드에서 가져오는 줄입니다.

```python
import pandas as pd
```

## 3. CSV를 DataFrame으로 읽습니다

긴 표를 코드 안에 옮겨 적지 않습니다. 파일 경로 한 줄로 읽습니다.

```python
df = pd.read_csv("../data/04-library-data.csv")
df.head()
```

## 4. 표의 기본 상태와 평균을 확인합니다

행과 열의 수, 열 이름, 첫 날짜와 두 평균을 원본과 맞춰 봅니다.

```python
print("행과 열:", df.shape)
print("열 이름:", list(df.columns))
print("첫 날짜:", df.loc[0, "date"])
print("방문자 평균:", round(df["visitors"].mean(), 2))
print("대출 권수 평균:", round(df["borrowed_books"].mean(), 2))
```

## 5. 열 이름으로 그래프를 그립니다

4-1에서는 AI가 Excel 값을 Matplotlib 코드에 옮겼습니다. pandas에서는 열 이름으로 바로 연결합니다.

```python
ax = df.plot(x="date", y="visitors", kind="line", legend=False,
             color="#FF5C00", figsize=(10, 4.5))
ax.set_title("Library Visitors")
ax.set_xlabel("Date")
ax.set_ylabel("Visitors")
positions = list(range(0, len(df), 5))
ax.set_xticks(positions)
ax.set_xticklabels(df["date"].iloc[positions], rotation=45, ha="right")
```

## 6. 5일 평균 추세선을 추가합니다

자료에는 주말이 없어서 5행이 한 주의 평일입니다. 하루 값은 회색으로 두고, 최근 5개 평일 평균을 주황 선으로 겹쳐 전체 흐름을 봅니다.

```python
df["visitor_trend"] = df["visitors"].rolling(window=5, min_periods=1).mean()

ax = df.plot(x="date", y="visitors", color="#8B9098",
             figsize=(10, 4.5), label="Daily visitors")
df.plot(x="date", y="visitor_trend", color="#FF5C00",
        linewidth=3, ax=ax, label="5-day average")
ax.set_title("Library Visitors and Trend")
ax.set_xlabel("Date")
ax.set_ylabel("Visitors")
positions = list(range(0, len(df), 5))
ax.set_xticks(positions)
ax.set_xticklabels(df["date"].iloc[positions], rotation=45, ha="right")
```

## 정리

- pandas는 CSV를 DataFrame으로 읽습니다.
- 머리글은 DataFrame의 열 이름이 됩니다.
- 열 이름을 사용하면 표와 그래프를 짧게 연결할 수 있습니다.
- 최근 5개 평일 평균을 겹치면 요일별 오르내림 속 전체 흐름을 보기 쉽습니다.
