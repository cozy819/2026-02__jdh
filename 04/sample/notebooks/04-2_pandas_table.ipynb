## 1. 3-2에서 만든 이름표 묶음

지난 시간에 한 사람을 이름표 묶음 하나로 적었습니다. 이번에는 **이름표마다 값이 여러 개**입니다.

````python
data = {
    "day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "visitors": [18, 25, 21, 32, 29],
}

print(data)
````

`print` 로는 한 줄로 길게 나옵니다. 이걸 **표 모양**으로 보고 싶습니다.

## 2. 표를 다루는 도구를 가져옵니다

## 3. dict 가 표가 됩니다

같은 `data` 입니다. 모양만 바뀝니다.

````python
import pandas as pd

df = pd.DataFrame(data)
df
````

**이름표가 열 이름이 되고, 값 하나하나가 행이 되었습니다.**
`df` 는 DataFrame 을 담은 이름입니다. pandas 는 도구 이름, DataFrame 은 그 도구가 만든 표입니다.

## 4. 표의 기본 상태를 확인합니다

표를 받으면 늘 이 세 가지를 먼저 봅니다.

````python
print("행과 열:", df.shape)
print("열 이름:", list(df.columns))
print("가장 큰 값:", df["visitors"].max())
````

## 5. 열 이름으로 그래프까지 잇습니다

4-1에서는 목록 두 개를 따로 넘겼습니다. pandas 는 **열 이름만 적으면** 됩니다.

````python
ax = df.plot(x="day", y="visitors", kind="bar", legend=False,
             color="#72757d", figsize=(8, 4.5))
ax.set_ylim(0, 36)
ax.set_title("Visitors by Day")
ax.set_xlabel("Day")
ax.set_ylabel("Visitors (people)")
labels = ax.bar_label(ax.containers[0])
````

## 6. 원본과 대조합니다

| 확인할 것 | 원본 |
|---|---|
| 행 수 | 5 |
| 열 이름 | day · visitors |
| 같은 행의 값 | Thu 는 32 |
| 가장 큰 값 | 32 |
| 막대 개수와 높이 | 5개 · 표와 같음 |
| 제목·축 이름·단위 | 있음 |

## 정리

- **pandas 는 도구, DataFrame 은 그 도구가 만든 표**입니다.
- dict 의 이름표가 표의 **열 이름**이 됩니다.
- 열 이름으로 표와 그래프를 짧게 이을 수 있습니다.
- 쉽게 그렸다고 확인이 없어지지는 않습니다. **행·열·값을 원본과 대조합니다.**
