# 4-1 Matplotlib으로 지정 그래프 확인하기
요청 조건과 원본 표를 읽은 뒤 셀을 실행합니다.

**요청 조건**
- 요일별 이용 인원을 막대그래프로 표시
- 가로축: 요일, 세로축: 이용 인원(명)
- 세로축은 0에서 시작
- 제목과 축 이름 표시

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
visitors = [18, 25, 21, 32, 29]
print(list(zip(days, visitors)))
```

```python
from pathlib import Path
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4.5))
bars = plt.bar(days, visitors, color=["#72757d", "#72757d", "#72757d", "#ff5a00", "#72757d"])
plt.ylim(0, 36)
plt.title("Visitors by Day")
plt.xlabel("Day")
plt.ylabel("Visitors (people)")
plt.bar_label(bars)
plt.tight_layout()
output = Path("../output/04-1_visitors_bar.png")
output.parent.mkdir(exist_ok=True)
plt.savefig(output, dpi=160)
plt.show()
```

## 확인
막대 5개, 값 18·25·21·32·29, 목요일 최대, 제목·축·단위, 세로축 0 시작을 원본과 대조합니다.
