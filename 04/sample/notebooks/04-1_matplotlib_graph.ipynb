## 4. 같은 코드로 그래프를 그립니다

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4.5))
bars = plt.bar(days, visitors, color="#72757d")
plt.ylim(0, 36)
plt.title("Visitors by Day")
plt.xlabel("Day")
plt.ylabel("Visitors")
plt.bar_label(bars)
plt.tight_layout()
plt.show()
````
