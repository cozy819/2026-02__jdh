# 4-1. Matplotlib으로 그래프 이해하기

셀을 **위에서 아래로** 하나씩 실행합니다. 코드는 고치지 않습니다.
중간에 **일부러 오류가 나는 셀**이 있습니다. 오류가 나야 정상입니다.

## 1. 오늘 그릴 자료

요일별 이용 인원입니다. 먼저 값부터 봅니다.

````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
visitors = [18, 25, 21, 32, 29]

print(days)
print(visitors)
````

## 2. 그래프를 그려 봅니다 — 여기서 멈춥니다

이 셀은 **오류가 나는 것이 정상**입니다. 긴 글은 다 읽지 말고 **마지막 줄만** 보세요.

````python
import matplotlib.pyplot as plt
````

마지막 줄에 이렇게 나옵니다.

```
ModuleNotFoundError: No module named 'matplotlib'
```

`No module named` — 그런 이름의 도구가 이 폴더에 없다는 뜻입니다.
코드가 틀린 것이 아니라 **쓸 도구를 아직 안 가져온 것**입니다.

## 3. 도구를 가져옵니다

터미널에 아래 한 줄을 칩니다. 그리고 **커널을 다시 시작**합니다.

```
uv add matplotlib
```

## 4. 다시 그려 봅니다

이번에는 됩니다. 같은 코드입니다.

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4.5))
bars = plt.bar(days, visitors, color="#72757d")
plt.ylim(0, 36)
plt.title("Visitors by Day")
plt.xlabel("Day")
plt.ylabel("Visitors (people)")
plt.bar_label(bars)
plt.tight_layout()
plt.show()
````

## 5. 원본과 대조합니다

그림이 나왔다고 끝이 아닙니다. **원본 값과 맞는지 눈으로 확인합니다.**

| 확인할 것 | 원본 |
|---|---|
| 막대 개수 | 5개 |
| 각 막대 값 | 18 · 25 · 21 · 32 · 29 |
| 가장 높은 요일 | Thu |
| 세로축 시작 | 0 |
| 제목·축 이름·단위 | 있음 |

## 6. 같은 숫자, 다른 인상

이번에는 **다른 주**의 이용 인원입니다. 다섯 날의 값이 서로 비슷합니다.

아래 두 셀은 **값이 똑같습니다.** 다른 것은 세로축이 어디서 시작하는지뿐입니다.
실행하기 전에 두 그림이 어떻게 다를지 예상해 보세요.


````python
next_week = [28, 30, 29, 33, 31]

plt.figure(figsize=(6, 4))
plt.bar(days, next_week, color="#72757d")
plt.ylim(0, 35)
plt.title("A")
plt.tight_layout()
plt.show()
````

````python
plt.figure(figsize=(6, 4))
plt.bar(days, next_week, color="#72757d")
plt.ylim(15, 35)
plt.title("B")
plt.tight_layout()
plt.show()
````

## 정리

- 라이브러리는 **이미 만들어진 기능의 묶음**입니다. 없으면 가져와야 합니다.
- `uv add`는 이 폴더에 도구를 챙겨 넣고, `import`는 코드에서 그 도구를 불러옵니다. **다른 일입니다.**
- 그래프가 나왔다고 맞는 것이 아닙니다. **원본과 대조해야** 맞는 것입니다.
- 세로축이 어디서 시작하는지에 따라 같은 숫자가 다르게 보입니다. A와 B가 그것입니다.
