## 2. 계산과 비교는 결과의 종류가 다르다

실행하기 전에 두 줄의 결과를 예상해 보세요.

```python
score = 72

print(score + 10)
print(score >= 60)
```

위는 **숫자**, 아래는 **True 또는 False**입니다. 계산은 값을 만들고 비교는 판단을 만듭니다.

## 3. 여섯 가지 비교

각 줄의 결과를 먼저 예상하고 실행하세요.

```python
print(8 > 5)
print(8 < 5)
print(8 >= 8)
print(8 <= 7)
print(8 == 8)
print(8 != 5)
```

## 4. 경계값 — 정확히 기준에 걸린 사람

`60점 이상`이라는 같은 기준에 59, 60, 61을 넣어 봅니다.

```python
print(59 >= 60)
print(60 >= 60)
print(61 >= 60)
```

가운데 줄이 이 차시의 핵심입니다. **정확히 60점인 사람도 통과입니다.**

`>` 였다면 결과가 달라집니다.

```python
print(60 > 60)
print(60 >= 60)
```

## 정리

- 계산의 결과는 숫자, 비교의 결과는 `True` 또는 `False`
- 등호가 붙어 있으면(`>=`, `<=`) 같을 때도 참
- `=`는 저장, `==`는 비교
