# 3-2. 여러 값을 묶는 두 가지 방법 — list와 dict

셀을 **위에서 아래로** 하나씩 실행합니다.
코드는 고치지 않습니다. 실행 결과만 읽습니다.

## 1. list — 순서로 묶기

값 네 개가 이름 하나에 들어 있습니다.

```python
scores = [72, 85, 90, 68]

print(scores)
```

## 2. 번호로 값 꺼내기

대괄호 안의 번호는 **맨 앞에서 몇 칸 떨어졌는지**를 뜻합니다.

```python
scores = [72, 85, 90, 68]

print(scores[0])
print(scores[1])
print(scores[2])
print(scores[3])
```

| 번호 | 뜻 | 값 |
|---|---|---|
| `0` | 맨 앞에서 0칸 | 72 |
| `1` | 맨 앞에서 1칸 | 85 |
| `2` | 맨 앞에서 2칸 | 90 |
| `3` | 맨 앞에서 3칸 | 68 |

값이 **4개**이면 마지막 번호는 **3**입니다. 개수보다 하나 작습니다.

## 3. 값이 몇 개인지 세어 본다

`len`은 **length**, 곧 길이입니다. 몇 개가 들어 있는지 세어 줍니다.


```python
scores = [72, 85, 90, 68]

print(len(scores))
```

값은 **4개**입니다. 자리 번호는 `0` `1` `2` `3` 이고 **4번 자리는 없습니다.**

아래 셀에서 일부러 그 없는 자리를 불러 봅니다.


## 4. 없는 자리를 부르면

이 셀은 **오류가 나는 것이 정상**입니다. 긴 글을 다 읽지 말고 **마지막 줄만** 보세요.

```python
scores = [72, 85, 90, 68]

print(scores[4])
```

마지막 줄에 `IndexError`와 `out of range`가 보입니다. `range`는 범위라는 뜻입니다.

**없는 자리를 부르면 프로그램이 멈춥니다.**

## 5. dict — 이름으로 찾기

번호 대신 이름표로 값을 꺼냅니다.

```python
student = {
    "name": "민준",
    "score": 85,
    "passed": True,
}

print(student)
print(student["name"])
print(student["score"])
```

## 6. 나를 이름표 묶음으로

워크시트에 적은 표와 이 코드의 모양을 견주어 보세요.

```python
me = {
    "name": "민준",
    "grade": 1,
    "club": "밴드",
    "favorite_subject": "음악",
}

print(me["club"])
print(me["favorite_subject"])
```

## 7. 그럼 여러 명은?

한 사람이 이름표 묶음 하나입니다. 같은 것이 여러 개 늘어서면 순서로 묶습니다.

```python
students = [
    {"name": "민준", "club": "밴드"},
    {"name": "서연", "club": "미술"},
]

print(students[0])
print(students[1]["club"])
```

문법을 외우지 않습니다. **이 모양이 곧 표 한 장**이라는 것만 확인합니다.

| name | club |
|---|---|
| 민준 | 밴드 |
| 서연 | 미술 |

이름표가 표의 **열 이름**이 되고, 묶음 하나가 표의 **한 행**이 됩니다.

## 정리

- list는 **순서**로, dict는 **이름**으로 값을 찾는다
- 번호 `0`은 첫 번째라는 뜻이 아니라 **맨 앞에서 0칸**이라는 뜻
- 값이 n개면 마지막 번호는 n-1, 범위를 넘으면 오류
- 이름표 묶음을 여러 개 모으면 표가 된다
